```python
import os
import json

# Check where your frames are (relative to T-DEED folder)
frame_dir = "extracted_frames448"
if os.path.exists(frame_dir):
    print("Frame directory exists:", frame_dir)
    # List the structure
    for root, dirs, files in os.walk(frame_dir):
        level = root.replace(frame_dir, '').count(os.sep)
        indent = ' ' * 2 * level
        print(f"{indent}{os.path.basename(root)}/")
        subindent = ' ' * 2 * (level + 1)
        for file in files[:5]:  # Show first 5 files
            print(f"{subindent}{file}")
        if len(files) > 5:
            print(f"{subindent}... and {len(files)-5} more files")
```
